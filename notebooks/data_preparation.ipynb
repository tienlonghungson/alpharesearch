```python
import yfinance as yf
import pandas as pd
import os

DATA_DIR = '../data'

test_tickers = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'NVDA']

def download_data(tickers, start_date = "2020-01-01", end_date = "2025-01-01"):
    print("Downloading data ...")
    data = yf.download(tickers, start = start_date, end = end_date)

    print(data.head(10))
    print("----------------------------------------\n")
    
    return data
    
data = download_data(test_tickers) ## yf.download is Wide Format MultiIndex dataframe
```
